```python
import numpy as np
from numpy import linalg as LA

#5. Compute the distance and angle between the two vectors

v1 = np.array([1, 2, 3])
v2 = np.array([-1, -1, 0])

# getting the distance using Linear Algebra
distance = LA.norm(v1 - v2)

# getting the dot product
dot_product = np.dot(v1, v2)

norm_v1 = LA.norm(v1)
norm_v2 = LA.norm(v2)

cosine_theta = dot_product / (norm_v1 *norm_v2)
arc_cos = np.arccos(cosine_theta)

# Print out the answers
print(f"Distance: {distance:.2f}")
print(f"Angle: {arc_cos:.2f}")

```
