```python
import numpy as np
from numpy import linalg as LA
import matplotlib.pyplot as plt

#7. Using Python, perform a principal component analysis on the dataset.

matrix = np.array([[5.0, 4.8],
                   [1.0, 1.4],
                   [4.4, 5.8],
                   [3.8, 4.4],
                   [6.2, 6.0],
                   [4.6, 5.4],
                   [4.0, 3.2],
                   [2.0, 2.2],
                   [3.0, 3.2],
                   [2.2, 1.8]])

#7a Visualize the dataset

#plt.scatter(matrix[:, 0], matrix[:, 1], color='red', label="dataset")
#plt.grid(True)
#plt.show()

#7b Normalize the data

mean = np.mean(matrix, axis=0)

matrix_centered = matrix - mean

#7c Compute the covariance

cov_matrix = np.cov(matrix_centered, rowvar = False)
print("Covariance Matrix:\n", cov_matrix)

#7d compute the eigenvalues and eigenvectors

eigenvalues, eigenvectors = LA.eig(cov_matrix)

#7e selecting the principal components

largest_index = np.argmax(eigenvalues)
principal_component = eigenvectors[:, largest_index]
matrix_reduced = np.dot(matrix_centered, principal_component)

#7f project the transformed data

matrix_reconstruct = np.dot(matrix_reduced.reshape(-1, 1), principal_component.reshape(1, -1)) + mean

#7g plot reconstructed data


plt.scatter(matrix_reconstruct[:,0], matrix_reconstruct[:, 1], color='blue', marker='x')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Original v. Reconstructed Data')

plt.grid(True)
plt.show()
```
