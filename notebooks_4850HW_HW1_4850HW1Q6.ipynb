```python
import numpy as np
from numpy import linalg as LA

#6. Use the numpy package to compute the SVD of the matrix.
# Also show the rank-1 approximation of the matrix using U sum and V of the SVD

A = np.array([[3, 2, 2],[2, 3, -2]], dtype=float)

# getting the SVD
U, s, Vh = np.linalg.svd(A)

print("Answers are:" U, s, Vh)
```
