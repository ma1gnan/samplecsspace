```python
import numpy as np
from numpy import linalg as LA
import matplotlib.pyplot as plt

matrix = np.array([[5.0, 4.8],
                   [1.0, 1.4],
                   [4.4, 5.8],
                   [3.8, 4.4],
                   [6.2, 6.0],
                   [4.6, 5.4],
                   [4.0, 3.2],
                   [2.0, 2.2],
                   [3.0, 3.2],
                   [2.2, 1.8]])

mean = np.mean(matrix, axis=0)

#8 

matrix_centered = matrix - np.mean(matrix, axis=0)

U, S_svd, VT = np.linalg.svd(matrix_centered)

V= VT.T
Vk = V[:, 0:1]
matrix_reduced_svd = np.dot(matrix_centered, Vk)
matrix_reconstructed_svd = np.dot(matrix_reduced_svd, Vk.T) + mean

plt.scatter(matrix[:, 0], matrix[:, 1], color='red')
plt.scatter(matrix_reconstructed_svd[:, 0], matrix_reconstructed_svd[:, 1], color='green', marker='X')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Original v. Reconstructed Data')
plt.grid(True)
plt.show()
```
